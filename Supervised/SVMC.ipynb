```python
import pandas as pd
import numpy as np
from sklearn.metrics import accuracy_score
from sklearn.svm import SVC
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

```python
df=pd.read_csv(r"C:\Users\User\OneDrive\Desktop\python\.newvenv\Data\student_performance.csv")
```

```python
df
```

```python
df.isnull().sum()
```

```python
#drop thecolumns that are not needed for the analysis
df = df.drop(['StudentID', 'Name', 'Gender', 'ExtracurricularActivities', 'ParentalSupport', 'Online Classes Taken'], axis=1)
```

```python
df.dropna()
```

```python
df.isnull().sum()
```

```python
#filna Nan Values
df.fillna(df.mean(), inplace=True)
```

```python
df.isnull().sum()
```

```python
df.head(1)
```

```python
#add extra column to the dataset to indicate whether the student passed or failed based on their final grade
df['Pass/Fail'] = np.where(df['FinalGrade'] >= 70, 'Pass', 'Fail')
```

```python
df
```

```python
X = df.drop(['Pass/Fail'], axis=1)
```

```python
y = df['Pass/Fail']
```

```python
#train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,random_state=42, test_size=0.2)
```

```python
#Scaling the data
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
#svr model
model=SVC(
    kernel='rbf',
    gamma='scale',
    )  
```

```python
model.fit(X_train,y_train)
```

```python
#pred
y_pred=model.predict(X_test)
```

```python
y_pred
```

```python
from sklearn.metrics import classification_report
```

```python
print(classification_report(y_test, y_pred, target_names=['Fail', 'Pass']))
```

```python
print(classification_report(y_test, y_pred))
```
