# Image classification demo
This notebook trains a simple classifier on scikit-learn's `digits` dataset.
Use the cells below to run training and evaluate accuracy.

```python
import numpy as np
from sklearn.datasets import load_digits
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report
import joblib
```

```python
def load_data():
    data = load_digits()
    X = data.images
    y = data.target
    n_samples = X.shape[0]
    X = X.reshape((n_samples, -1))
    return X, y

def get_classifier(name):
    if name == 'svm':
        return SVC(kernel='rbf', gamma=0.001, C=1.0)
    if name == 'rf':
        return RandomForestClassifier(n_estimators=100, random_state=42)
    raise ValueError(f'Unknown model: {name}')
```

```python
# Configure and run training
model_name = 'svm'  # choose 'svm' or 'rf'
test_size = 0.2
save_path = ''  # e.g. 'model.joblib' to save

X, y = load_data()
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=test_size, random_state=42, stratify=y
)
clf = get_classifier(model_name)
clf.fit(X_train, y_train)
preds = clf.predict(X_test)
print(f'Test accuracy: {accuracy_score(y_test, preds):.4f}')
print('Classification report:')
print(classification_report(y_test, preds))
if save_path:
    joblib.dump(clf, save_path)
    print(f'Saved model to {save_path}')
```
