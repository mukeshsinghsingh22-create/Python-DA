```python
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.cluster import KMeans
```

```python
df = pd.read_csv(r"E:\ML\data\data_set.csv")

print(df.head())
```

```python
X = df[["Annual_Income_k$", "Spending_Score"]]

X.head()
```

```python
model = KMeans(
    n_clusters=3,
    random_state=42
)
```

```python
model.fit(X)
```

```python
clusters = model.predict(X)

print(clusters)
```

```python
df["Cluster"] = clusters

print(df.head())
```

```python
plt.scatter(
    df["Annual_Income_k$"],
    df["Spending_Score"],
    c=df["Cluster"]
)

plt.xlabel("Annual Income")
plt.ylabel("Spending Score")
plt.title("Customer Clusters")

plt.show()
```

```python
print(model.cluster_centers_)
```

```python
plt.scatter(
    df["Annual_Income_k$"],
    df["Spending_Score"],
    c=df["Cluster"]
)

plt.scatter(
    model.cluster_centers_[:,0],
    model.cluster_centers_[:,1],
    color="red",
    marker="X",
    s=250
)

plt.xlabel("Annual Income")
plt.ylabel("Spending Score")
plt.title("Customer Clusters")

plt.show()
```
